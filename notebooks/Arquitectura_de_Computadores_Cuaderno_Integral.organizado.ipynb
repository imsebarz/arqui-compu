### Sistemas de numeración
Un sistema posicional representa un número como suma de dígitos multiplicados por potencias de la base. Usamos decimal (10), binario (2), octal (8) y hexadecimal (16). ASCII asocia caracteres con números; por ejemplo, una palabra como UDEA se almacena como códigos binarios.

Ejemplos del apunte: 65 y 745 de decimal a binario; 879 a octal; 1352 a hexadecimal; además conversiones inversas desde binario y hexadecimal.

### Operaciones binarias

- **0 + 0 = 0** (llevo 0)

- **1 + 1 = 0** y **llevo 1** (porque 2 en decimal = 10 en binario)
- Si hay “llevada” (carry), también se suma: por ejemplo **1 + 1 + 1 = 1** y **llevo 1** (porque 3 = 11)

## 2. Codificación de Huffman

<div style="border-left: 4px solid #0f766e; background: #f0fdfa; padding: 12px 16px; margin: 8px 0 18px 0; border-radius: 4px;">
  <strong>Objetivo</strong><br>Construir un árbol de Huffman con una regla de desempate explícita y verificar que la codificación se decodifica sin pérdida.<br><br>
  <strong>Incluye</strong><br>Frecuencias, cola de prioridad, árbol, código prefijo, ejercicio completo, calculadora y métricas de compresión.

Huffman es un método de compresión **sin pérdida** que asigna códigos cortos a símbolos frecuentes y largos a símbolos raros. El código resultante es de **prefijo**: ningún código completo es prefijo de otro, por lo que se decodifica sin separadores.

### Algoritmo
1. Normalizar el texto según las reglas del ejercicio.
2. Contar frecuencias.
3. Crear un nodo por símbolo.
4. Repetir: extraer los dos nodos de menor frecuencia y unirlos.
5. Etiquetar ramas y recorrer el árbol hasta cada hoja.
6. Sustituir cada símbolo por su código.
7. Verificar que al decodificar se recupera el texto.

En el apunte: desempates en orden alfabético Z→A; se asigna `1` al nodo de menor frecuencia y `0` al de mayor frecuencia. La calculadora mantiene una regla determinista equivalente y muestra todos los pasos.


#### 1) ¿Qué problema resuelve?

Comprime sin pérdida(puedes recuperar el mensaje exacto) usando códigos de longitud variable:

- símbolos que aparecen mucho → pocos bits
- símbolos raros → más bits

La meta es minimizar la longitud promedio del mensaje codificado.

---


#### 2) Idea clave: código prefijo

Huffman construye un código prefijo: ningún código es el inicio de otro.

Ejemplo válido: A=0, B=10, C=110, D=111

Aquí, cuando lees bits de izquierda a derecha, sabes exactamente cuándo termina un símbolo (porque no existe un símbolo cuyo código sea prefijo del siguiente).

Esto se logra porque los códigos salen de un árbol binario:

- cada símbolo es una hoja
- izquierda = 0, derecha = 1
- el código es el camino desde la raíz hasta la hoja

---


#### 3) Algoritmo (paso a paso)

Entrada:tabla de frecuencias (o probabilidades) por símbolo.

1. Crea un nodo por símbolo con su frecuencia.
2. Ponlos en una cola de prioridad (min-heap)por frecuencia.
3. Mientras haya más de un nodo:
4. Al final queda un nodo: la raíz.
5. Etiqueta ramas (convención del profe): **1** para el nodo de **menor** frecuencia y **0** para el de **mayor** frecuencia (en cada unión).
6. Recorre el árbol: el camino define el código de cada símbolo.

### 5) ¿Cómo se codifica y decodifica?


Codificar: reemplazas símbolos por bits y concatenas.

### 🧪 Implementación reproducible


````python
from collections import Counter
import heapq, itertools, unicodedata

def normalizar(texto, contar_espacios=False, ignorar_tildes=True):
    if ignorar_tildes:
        texto = ''.join(c for c in unicodedata.normalize('NFD', texto) if unicodedata.category(c) != 'Mn')
    texto = texto.upper()
    return texto if contar_espacios else texto.replace(' ', '')

def huffman(texto, contar_espacios=False):
    limpio = normalizar(texto, contar_espacios)
    frecuencias = Counter(limpio)
    contador = itertools.count()
    heap=[]
    # desempate reproducible: símbolo inverso para aproximar Z→A
    for simbolo, frecuencia in sorted(frecuencias.items(), reverse=True):
        heapq.heappush(heap, (frecuencia, next(contador), simbolo))
    pasos=[]
    if len(heap)==1:
        return limpio, frecuencias, {heap[0][2]:'0'}, [], heap[0][2]
    while len(heap)>1:
        f1, _, n1 = heapq.heappop(heap)
        f2, _, n2 = heapq.heappop(heap)
        nuevo=(n1,n2)  # izquierda=1 (menor/primero), derecha=0
        pasos.append((n1,f1,n2,f2,f1+f2))
        heapq.heappush(heap,(f1+f2,next(contador),nuevo))
    raiz=heap[0][2]; codigos={}
    def recorrer(nodo,prefijo=''):
        if isinstance(nodo,str): codigos[nodo]=prefijo or '0'; return
        recorrer(nodo[0],prefijo+'1'); recorrer(nodo[1],prefijo+'0')
    recorrer(raiz)
    return limpio, frecuencias, codigos, pasos, raiz

def codificar(texto,codigos): return ''.join(codigos[c] for c in texto)

def decodificar(bits,raiz):
    if isinstance(raiz,str): return raiz*len(bits)
    salida=[]; nodo=raiz
    for bit in bits:
        nodo=nodo[0] if bit=='1' else nodo[1]
        if isinstance(nodo,str): salida.append(nodo); nodo=raiz
    if nodo is not raiz: raise ValueError('Secuencia incompleta')
    return ''.join(salida)
````

````python
texto='ARQUITECTURA DE COMPUTADORES'
limpio, frecs, codigos, pasos, arbol = huffman(texto)
print('Frecuencias:', dict(sorted(frecs.items())))
print('\nUniones:')
for i,(a,fa,b,fb,total) in enumerate(pasos,1): print(f'{i:2}. {a}({fa}) + {b}({fb}) = {total}')
print('\nTabla de códigos:')
for s in sorted(codigos): print(f'{s}: {frecs[s]} → {codigos[s]}')
bits=codificar(limpio,codigos)
print('\nTexto normalizado:',limpio)
print('Bits:',bits)
print('Longitud:',len(bits),'bits; fijo de 4 bits:',len(limpio)*4,'bits')
print('Decodifica correctamente:',decodificar(bits,arbol)==limpio)
````

Un código fijo usa `ceil(log2(m))` bits por símbolo; Huffman usa `suma(frecuencia × longitud)`. Verifica suma de frecuencias, propiedad de prefijo, decodificación exacta y ahorro frente al código fijo.


````python
import math
def metricas_huffman(texto):
    limpio, frecs, codigos, pasos, arbol=huffman(texto)
    total=len(limpio); bh=sum(frecs[s]*len(codigos[s]) for s in frecs)
    bf=total*max(1,math.ceil(math.log2(len(frecs))))
    prefijo=all(not codigos[a].startswith(codigos[b]) for a in codigos for b in codigos if a!=b)
    return {'bits_huffman':bh,'bits_fijos':bf,'longitud_media':bh/total,'ahorro_%':100*(1-bh/bf),'es_prefijo':prefijo}
print(metricas_huffman('ARQUITECTURA DE COMPUTADORES'))
````

### Números negativos en binario

**Para representar números con signo se fija un número de bits.** Con **n** bits, el bit más significativo (MSB) suele indicar el signo: **0** positivo y **1** negativo.


#### 1. Signo y magnitud

El primer bit representa el signo y los demás almacenan el valor absoluto.

- Ejemplo con 8 bits: +13 = **00001101**
- −13 = **10001101**
- Rango con 8 bits: de −127 a +127.
- Problema: existen dos ceros: +0 = 00000000 y −0 = 10000000.

#### 2. Complemento a 1 (C1)

Un número negativo se obtiene invirtiendo cada bit de su representación positiva: 0 ↔ 1.

- +13 = **00001101**
- −13: invertir los bits → **11110010**
- Rango con 8 bits: de −127 a +127.
- También tiene dos ceros: +0 = 00000000 y −0 = 11111111. 859
- 512 256 128 64 32 16 8 4 2 1
- 1 1 0 1 0 1 1 0 1 1
- 1101011011 -> 0010100100 en complemento a 1

#### 3. Complemento a 2 (C2)

Es el método más usado. Para obtener el negativo: primero se calcula el complemento a 1 y luego se suma 1.

- +13 = **00001101**
- C1 de 13 = **11110010**
- C2 de −13 = 11110010 + 1 = **11110011**

#### Métodos para obtener el complemento a 2

**Importante:**

todos los métodos dan el mismo resultado y se debe conservar siempre la misma cantidad de bits.

1. **Método 1: complemento a 1 + 1.** Invierte todos los bits y suma 1 al resultado.
2. **Método 2: atajo de derecha a izquierda.** Desde la derecha, conserva los bits hasta encontrar el primer 1; conserva también ese 1 e invierte todos los bits restantes hacia la izquierda. Este atajo equivale a invertir y sumar 1, incluidos los acarreos.
3. **Método 3: resta modular.** Con n bits, −x se puede calcular como 2<sup>n</sup> − x. Por ejemplo, con 8 bits: −76 = 256 − 76 = 180 = 10110100₂.

### Formato de coma flotante en binario


El **formato de coma flotante binario** permite representar números con parte decimal, muy grandes o muy pequeños. Se expresa de forma parecida a la notación científica:

**± mantisa × 2<sup>exponente</sup>**

Por ejemplo:

**196,3<sub>10</sub> = 11000100,0100110011…<sub>2</sub> = 1,10001000100110011…<sub>2</sub> × 2<sup>7</sup>**

Se compone de tres campos:

- **Signo:** indica si el número es positivo (0) o negativo (1).
- **Exponente:** indica cuánto se desplaza la coma binaria.
- **Mantisa o fracción:** guarda los dígitos significativos del número.

Este formato permite un rango amplio de valores, aunque algunos decimales no pueden representarse exactamente en binario y se aproximan.

#### Conversión de 196,3

<sub>10</sub> a binario

Se separa el número en una parte entera y una parte fraccionaria:

**196,3 = 196 + 0,3**

##### 1. Parte entera: 196

Se realizan divisiones sucesivas entre 2 y se anotan los residuos.

| División | Cociente | Residuo |
| --- | --- | --- |
| 196 ÷ 2 | 98 | 0 |
| 98 ÷ 2 | 49 | 0 |
| 49 ÷ 2 | 24 | 1 |
| 24 ÷ 2 | 12 | 0 |
| 12 ÷ 2 | 6 | 0 |
| 6 ÷ 2 | 3 | 0 |
| 3 ÷ 2 | 1 | 1 |
| 1 ÷ 2 | 0 | 1 |

Los residuos se leen de abajo hacia arriba:

**196<sub>10</sub> = 11000100<sub>2</sub>**

##### 2. Parte fraccionaria: 0,3

Se multiplica repetidamente la fracción por 2. La parte entera de cada resultado forma el siguiente bit después de la coma.

| Operación | Bit obtenido | Nueva fracción |
| --- | --- | --- |
| 0,3 × 2 = 0,6 | 0 | 0,6 |
| 0,6 × 2 = 1,2 | 1 | 0,2 |
| 0,2 × 2 = 0,4 | 0 | 0,4 |
| 0,4 × 2 = 0,8 | 0 | 0,8 |
| 0,8 × 2 = 1,6 | 1 | 0,6 |
| 0,6 × 2 = 1,2 | 1 | 0,2 |

Los bits obtenidos son

**010011…**

. A partir de 0,6 vuelven a aparecer las mismas fracciones:

**0,6 → 0,2 → 0,4 → 0,8 → 0,6**

. Por eso se repite el bloque

**1001**

:

**0,3<sub>10</sub> = 0,0100 1100 1100…<sub>2</sub> = 0,0100(1100)…<sub>2</sub>**

De forma compacta:

**0,3<sub>10</sub> = 0,010011001100…<sub>2</sub>**

.

##### ¿Por qué aparece la periodicidad?

Una fracción decimal tiene una representación binaria

**finita**

solo si, una vez simplificada, su denominador contiene únicamente factores 2. Por ejemplo, 0,75 = 3/4 y 4 = 2

<sup>2</sup>

, por eso termina en binario: 0,11

<sub>2</sub>

.

En cambio,

**0,3 = 3/10**

. Como 10 = 2 × 5 contiene el factor 5, no puede escribirse con una cantidad finita de bits binarios. El algoritmo de multiplicar por 2 nunca llega a fracción 0 y termina repitiendo fracciones; por ello la expansión es periódica.

En IEEE 754 solo hay un número fijo de bits para la fracción. Por tanto, una secuencia periódica como la de 0,3 debe

**redondearse**

, y el valor almacenado es una aproximación.

##### 3. Unir las dos partes

**196,3<sub>10</sub> = 11000100,010011001100…<sub>2</sub>**

#### Normalización para coma flotante

Se mueve la coma siete posiciones hacia la izquierda hasta dejar un solo 1 antes de ella:

**11000100,010011001100…<sub>2</sub> = 1,1000100010011001100…<sub>2</sub> × 2<sup>7</sup>**

- **Signo:** 0, porque el número es positivo.
- **Mantisa:** 1,1000100010011001100…<sub>2</sub>.
- **Exponente:** 7.

> Si el número fuera negativo, por ejemplo −196,3, la conversión binaria y la mantisa serían las mismas, pero el bit de signo sería 1 .

### Estándar IEEE 754


El estándar **IEEE 754** define cómo se representan los números con coma flotante en la mayoría de los computadores. Divide el número en tres campos:

**signo**, **exponente** y **fracción** o **mantisa**.

#### Formatos más comunes

| Formato | Total de bits | Signo | Exponente | Fracción |
| --- | --- | --- | --- | --- |
| Precisión simple | 32 | 1 bit | 8 bits | 23 bits |
| Precisión doble | 64 | 1 bit | 11 bits | 52 bits |

#### IEEE 754 de 32 bits

En precisión simple, los 32 bits se distribuyen así:

**S | EEEEEEEE | FFFFFFFFFFFFFFFFFFFFFFF**

- **S (signo):** vale 0 para números positivos y 1 para negativos.
- **E (exponente):** se almacena con un desplazamiento llamado **sesgo** o **bias**. En 32 bits, el sesgo es **127**.
- **F (fracción):** contiene los bits después de la coma de la mantisa.

#### Valor de un número normalizado

Cuando el exponente no es todo ceros ni todo unos, el valor se calcula así:

**(−1)<sup>S</sup> × 1,F × 2<sup>(E − 127)</sup>**

El

**1**

antes de la coma no se guarda: se asume de forma implícita. Por eso la mantisa se interpreta como

**1,F**

.

#### Ejemplo: representar 196,3 en IEEE 754 de 32 bits

La forma normalizada es:

**196,3<sub>10</sub> = 1,1000100010011001100…<sub>2</sub> × 2<sup>7</sup>**

1. **Signo:** el número es positivo, por tanto **S = 0**.
2. **Exponente real:** es 7.
3. **Exponente almacenado:** se suma el sesgo: 7 + 127 = 134. En binario, **134<sub>10</sub> = 10000110<sub>2</sub>**.
4. **Fracción:** se eliminan el 1 inicial implícito y la coma. Los primeros bits son **10001000100110011001100…**.
5. **Redondeo:** IEEE 754 de precisión simple conserva 23 bits de fracción. Los primeros 23 bits son **10001000100110011001100** y el bit siguiente es 1; al redondear al más cercano, queda **10001000100110011001101**.

La representación final es:

**0 | 10000110 | 10001000100110011001101**

Como la fracción binaria es infinita y periódica, este patrón representa una aproximación de 196,3, no el decimal exacto.

#### Casos especiales

- **E = 00000000 y F = 0:** representa +0 o −0, según el bit de signo.
- **E = 00000000 y F ≠ 0:** representa números muy pequeños, llamados **subnormales**.
- **E = 11111111 y F = 0:** representa +∞ o −∞.
- **E = 11111111 y F ≠ 0:** representa **NaN** (*Not a Number*), usado para resultados no numéricos, como 0/0.

> Idea clave: IEEE 754 no almacena directamente el número decimal. Guarda su signo, la mantisa binaria normalizada y un exponente con sesgo.

#### Proceso inverso: de IEEE 754 a decimal

Para recuperar un número decimal a partir de un patrón IEEE 754 de precisión simple (32 bits), se separan sus tres campos:

**S | EEEEEEEE | FFFFFFFFFFFFFFFFFFFFFFF**

1. **Signo:** determina si el valor es positivo o negativo.
2. **Exponente:** se convierte a decimal y se resta el sesgo 127.
3. **Fracción:** se reconstruye la mantisa como **1,F** para los números normalizados.
4. Se aplica la fórmula: **(−1)<sup>S</sup> × 1,F × 2<sup>(E − 127)</sup>**.

##### Ejemplo: 0 | 10001000 | 01001001000101000111101

##### 1. Signo

**S = 0**

, por tanto el número es positivo:

**(−1)<sup>0</sup> = +1**

##### 2. Exponente

El campo del exponente es

**10001000<sub>2</sub>**

.

**10001000<sub>2</sub> = 136<sub>10</sub>**

Se resta el sesgo de precisión simple:

**e = 136 − 127 = 9**

##### 3. Mantisa

La fracción almacenada es:

**F = 01001001000101000111101**

Como el exponente no es todo ceros ni todo unos, el número está normalizado. Por ello se agrega el

**1 implícito**

a la izquierda de la coma:

**1,F = 1,01001001000101000111101<sub>2</sub>**

Su valor decimal se obtiene sumando las potencias de 2 correspondientes a los bits con valor 1:

**1 + 2<sup>−2</sup> + 2<sup>−5</sup> + 2<sup>−8</sup> + 2<sup>−12</sup> + 2<sup>−16</sup> + 2<sup>−18</sup> + 2<sup>−19</sup> + 2<sup>−20</sup> + 2<sup>−21</sup> + 2<sup>−23</sup>**

**1,F = 1,2854686975…<sub>10</sub>**

##### 4. Aplicar el exponente

Ahora se multiplica por 2

<sup>9</sup>

:

**(−1)<sup>0</sup> × 1,2854686975… × 2<sup>9</sup>**

**= 1,2854686975… × 512**

**= 658,15997314453125**

Por tanto:

**0 | 10001000 | 01001001000101000111101 = 658,15997314453125<sub>10</sub>**

> El valor original 658,16 no se guarda exactamente: IEEE 754 almacena el valor binario representable más cercano, 658,15997314453125. La diferencia se debe a que 0,16 es periódico en binario.

##### Regla rápida

- Si **S = 0**, el número es positivo; si **S = 1**, es negativo.
- Para números normalizados: **exponente real = E − 127**.
- La mantisa siempre comienza por **1,**, aunque ese 1 no esté almacenado.
- Finalmente, se desplaza la coma binaria según el exponente o se multiplica por 2<sup>e</sup>.

### Código BCD


En precisión simple: 1 bit de signo, 8 de exponente con sesgo 127 y 23 de fracción. Para convertir: pasa a binario, normaliza como `1.f × 2^e`, guarda `e+127` y completa la mantisa. Las fracciones como 0.3 o 0.16 suelen ser periódicas en binario, por lo que se redondean. El apunte desarrolla 196.3 y 658.16 y la conversión inversa.


````python
import struct
def ieee754_32(x):
    bits=''.join(f'{b:08b}' for b in struct.pack('>f',x))
    return {'signo':bits[0], 'exponente':bits[1:9], 'mantisa':bits[9:], 'completo':bits, 'hex':struct.pack('>f',x).hex().upper()}
for x in (196.3,658.16): print(x,ieee754_32(x))
````

### BCD 8421 y display de 7 segmentos
Cada dígito decimal se codifica por separado con pesos 8-4-2-1: 0=`0000`, …, 9=`1001`. Los patrones 1010–1111 no representan dígitos decimales válidos. Esto explica por qué BCD no es igual a convertir el número completo a binario. Un decodificador BCD–7 segmentos activa `a…g` para mostrar cada dígito.


El **código BCD** (*Binary-Coded Decimal*, o decimal codificado en binario) representa cada **dígito decimal** de forma independiente mediante un grupo de 4 bits.

En BCD natural, también llamado

**8421**

, los pesos de los cuatro bits son 8, 4, 2 y 1.

#### Tabla de conversión BCD 8421

| Decimal | BCD |
| --- | --- |
| 0 | 0000 |
| 1 | 0001 |
| 2 | 0010 |
| 3 | 0011 |
| 4 | 0100 |
| 5 | 0101 |
| 6 | 0110 |
| 7 | 0111 |
| 8 | 1000 |
| 9 | 1001 |

#### Cómo convertir de decimal a BCD

Se separa el número decimal en sus dígitos y se sustituye cada uno por su grupo de 4 bits. No se convierte el número completo como si fuera un único número binario.

**Ejemplo: 593<sub>10</sub>**

- 5 → **0101**
- 9 → **1001**
- 3 → **0011**

Por tanto:

**593<sub>10</sub> = 0101 1001 0011<sub>BCD</sub>**

#### Diferencia entre binario puro y BCD

El número decimal 593 en binario puro es:

**593<sub>10</sub> = 1001010001<sub>2</sub>**

En cambio, en BCD se representa cada cifra decimal por separado:

**593<sub>10</sub> = 0101 1001 0011<sub>BCD</sub>**

Ambas representaciones corresponden al mismo valor decimal, pero usan reglas distintas.

#### Combinaciones no válidas

Con 4 bits se pueden formar 16 combinaciones, desde 0000 hasta 1111. Sin embargo, en decimal solo existen 10 dígitos, de 0 a 9.

- Combinaciones válidas: **0000** a **1001**.
- Combinaciones no válidas en BCD 8421: **1010** a **1111**.

> Idea clave: BCD facilita mostrar números decimales en dispositivos como relojes, calculadoras y pantallas, porque cada grupo de 4 bits representa directamente una cifra decimal.

#### BCD y display de 7 segmentos

##### Identificación de los segmentos

Los segmentos se nombran normalmente así:

##### Relación entre BCD y el display

**decodificador BCD a 7 segmentos**

Primero se expresa cada cifra en BCD:

##### Entradas y salidas

##### Esquema general

**Entradas BCD (A, B, C, D) → compuertas lógicas → salidas a, b, c, d, e, f, g → display**

Las compuertas NOT generan A̅, B̅, C̅ y D̅. Luego, las compuertas AND forman las combinaciones necesarias y una compuerta OR reúne las combinaciones que deben encender cada segmento.

**a = A̅B̅C̅D̅ + A̅B̅CD̅ + A̅B̅CD + A̅BC̅D + A̅BCD̅ + A̅BCD + AB̅C̅D̅ + AB̅C̅D**

### BCD validado
Separa en grupos de cuatro bits. Sólo `0000`–`1001` son dígitos; `1010`–`1111` son inválidos. En una suma BCD, si un grupo supera 9 o produce acarreo, se suma `0110`. Documenta si el display es ánodo común (activo en 0) o cátodo común (activo en 1).


````python
def validar_bcd(cadena):
    x=cadena.replace(' ','')
    if len(x)%4 or any(c not in '01' for c in x): return False,'Formato inválido'
    grupos=[x[i:i+4] for i in range(0,len(x),4)]; malos=[g for g in grupos if int(g,2)>9]
    return (not malos, malos or [int(g,2) for g in grupos])
for x in ('0010 0000 0010 0110','1001 0101','1010 0001'): print(x,'→',validar_bcd(x))
````

### Fundamentos de electrónica básica


#### ¿Qué es un circuito eléctrico?

Un circuito es un conjunto de componentes interconectados que permite el flujo de corriente eléctrica. Para funcionar, normalmente necesita una fuente de energía, conductores y una carga o componente que utilice la energía.

#### Magnitudes fundamentales

**Corriente eléctrica (I) -**

Se mide en

**amperios (A)**

Es el flujo de electrones a través de un circuito. Se mide en

**amperios (A)**

. Para medirla, el multímetro debe conectarse

**en serie**

, de modo que la corriente pase por el instrumento.

**Voltaje o tensión (V) -**

Se mide en

**voltios (V)**

Es la diferencia de potencial eléctrico entre dos puntos; puede entenderse como la fuerza que impulsa a los electrones. Se mide en

**voltios (V)**

. Las baterías proporcionan voltaje, por ejemplo, una pila de 9 V.

**Resistencia (R) -**

Se mide en

**ohmios (Ω)**

Es la oposición que presenta un material al paso de la corriente eléctrica. Se mide en

**ohmios (Ω)**

. Las resistencias se usan para limitar la corriente y proteger componentes.

#### Ley de Ohm

La relación entre voltaje, corriente y resistencia se expresa así:

$$
V = I × R
$$

De esta ecuación se obtienen también:

**I = V / R**

**R = V / I**

.

#### Conexiones básicas

- **Serie:** los componentes se conectan uno después de otro. La corriente es la misma en todos los elementos.
- **Paralelo:** los componentes se conectan en ramas independientes. El voltaje es el mismo en cada rama.
- **Mixto:** combina conexiones en serie y en paralelo.

#### Uso básico del multímetro

- Para medir **voltaje**, conectar el multímetro en paralelo con el componente o la fuente.
- Para medir **corriente**, abrir el circuito y conectar el multímetro en serie.
- Para medir **resistencia**, desconectar la alimentación del circuito antes de realizar la medición.

> Antes de energizar un circuito, verificar las conexiones, la polaridad de la fuente y el valor de los componentes.

#### Verificación y detección de fallas

Para localizar una falla, se debe seguir el recorrido de la corriente y comprobar hasta qué punto llega. Comparar las mediciones reales con los cálculos teóricos ayuda a identificar conexiones incorrectas, componentes dañados o valores equivocados.

### Notas de cálculo: Ley de Ohm, serie y paralelo


#### Ley de Ohm

Relaciona voltaje (

**V**

), corriente (

**I**

) y resistencia (

**R**

):

$$
V = I × R I = V / R R = V / I
$$

#### Circuitos en serie

Los componentes se conectan uno después de otro. La

**corriente es igual**

en todos los elementos y el voltaje se reparte entre ellos.

$$
Rₜ = R₁ + R₂ + R₃ + … Iₜ = I₁ = I₂ = I₃ Vₜ = V₁ + V₂ + V₃ + …
$$

#### Circuitos en paralelo

Los componentes se conectan en ramas independientes. El

**voltaje es igual**

en cada rama y la corriente total es la suma de las corrientes.

$$
1 / Rₜ = 1 / R₁ + 1 / R₂ + 1 / R₃ + … Vₜ = V₁ = V₂ = V₃ Iₜ = I₁ + I₂ + I₃ + …
$$

#### Ejemplo breve

Si V = 5 V y R = 2 400 Ω:

$$
I = V / R = 5 V / 2 400 Ω = 0,00208 A = 2,08 mA
$$

**Importante:**

usar siempre las mismas unidades: 2,2 kΩ = 2 200 Ω y 5 mA = 0,005 A.

Por ejemplo, 2,2 kΩ = 2 200 Ω y 5 mA = 0,005 A.

$$

Conectar las tres resistencias una después de la otra. La corriente sale del terminal positivo de la batería, pasa por R₁, R₂ y R₃, y regresa al terminal negativo.

#### Cálculos del circuito en serie (9 V)

Datos: Vₜ = 9 V

R₁ = 2,2 kΩ = 2 200 Ω - R₂ = 1,5 kΩ = 1 500 Ω - R₃ = 10 Ω

### 🧪 Verificación de circuitos con Python


Proceso: **calcular → diseñar → construir → medir**. Magnitudes: corriente (A), voltaje (V) y resistencia (Ω). Ley de Ohm: `V=IR`. En serie se suma resistencia; en paralelo se suman conductancias.

Instrumentos y seguridad: multímetro en paralelo para voltaje y en serie para corriente; seleccionar escala y bornes correctos; no medir resistencia con el circuito energizado. El montaje usa protoboard, resistores (código de colores), LED, jumpers, batería y eventualmente Arduino.

Ejemplo del apunte: fuente 9 V; R1=2.2 kΩ, R2=1.5 kΩ, R3=10 Ω. Total 3710 Ω, corriente ≈2.43 mA; caídas ≈5.34 V, 3.64 V y 0.02 V.


````python
def serie(voltaje,*resistencias):
    rt=sum(resistencias); i=voltaje/rt
    return rt,i,[i*r for r in resistencias]
rt,i,caidas=serie(9,2200,1500,10)
print(f'Rt={rt} Ω; I={i*1000:.3f} mA; caídas={[round(v,3) for v in caidas]}; suma={sum(caidas):.3f} V')
def paralelo(*resistencias): return 1/sum(1/r for r in resistencias)
print('Ejemplo paralelo 1kΩ || 2.2kΩ =',round(paralelo(1000,2200),2),'Ω')
````

### Circuitos ampliados
**Serie:** misma corriente. **Paralelo:** mismo voltaje. **Mixto:** reduce bloques sucesivamente. Potencia: `P=VI=I²R=V²/R`. Para LED: `R=(Vfuente−VLED)/ILED`. Mide resistencia sin energía, voltaje en paralelo y corriente insertando el multímetro en serie; comienza por la escala más alta.


````python
def circuito_paralelo(v,*rs):
    req=paralelo(*rs); ramas=[v/r for r in rs]
    return {'Req_ohm':req,'I_total_A':v/req,'I_ramas_A':ramas,'potencias_W':[v*v/r for r in rs]}
def resistor_led(v,vl,i_mA):
    r=(v-vl)/(i_mA/1000); return {'R_min_ohm':r,'P_W':(i_mA/1000)**2*r}
print(circuito_paralelo(9,1000,2200))
print(resistor_led(9,2,15))
````

### 5.1 Fundamentos y tablas de verdad



- BUFFER: `Y=A`
- NOT: `Y=¬A`
- AND: `Y=A·B`
- OR: `Y=A+B`
- NAND: `Y=¬(A·B)`
- NOR: `Y=¬(A+B)`
- XOR: `Y=A⊕B` (entradas diferentes)
- XNOR: `Y=¬(A⊕B)` (entradas iguales)

NAND y NOR son universales: con una sola de ellas se pueden construir NOT, AND y OR, y por tanto cualquier función booleana.


### Tabla de compuertas lógicas básicas


Convención:

**0**

= nivel bajo y

**1**

= nivel alto. A y B son entradas; Y es la salida.

| Compuerta | Tabla de verdad | Función | Diseño de la compuerta |
| --- | --- | --- | --- |
| BUFFER | A \| Y<br>0 \| 0<br>1 \| 1 | Y = A |   |
| NOT (Inversor) | A \| Y<br>0 \| 1<br>1 \| 0 | Y = ¬A |   |
| AND | A B \| Y<br>0 0 \| 0<br>0 1 \| 0<br>1 0 \| 0<br>1 1 \| 1 | Y = A · B |   |
| OR | A B \| Y<br>0 0 \| 0<br>0 1 \| 1<br>1 0 \| 1<br>1 1 \| 1 | Y = A + B |   |
| NAND | A B \| Y<br>0 0 \| 1<br>0 1 \| 1<br>1 0 \| 1<br>1 1 \| 0 | Y = ¬(A · B) |   |
| NOR | A B \| Y<br>0 0 \| 1<br>0 1 \| 0<br>1 0 \| 0<br>1 1 \| 0 | Y = ¬(A + B) |   |
| XOR (OR exclusiva) | A B \| Y<br>0 0 \| 0<br>0 1 \| 1<br>1 0 \| 1<br>1 1 \| 0 | Y = A ⊕ B = ¬A·B + A·¬B |   |
| XNOR (Equivalencia) | A B \| Y<br>0 0 \| 1<br>0 1 \| 0<br>1 0 \| 0<br>1 1 \| 1 | Y = ¬(A ⊕ B) = A·B + ¬A·¬B |   |


````python
from itertools import product
def tabla_compuertas():
    print('A B | AND OR NAND NOR XOR XNOR')
    for A,B in product((0,1),repeat=2):
        AND=A&B; OR=A|B; XOR=A^B
        print(A,B,'| ',AND,' ',OR,'  ',1-AND,'  ',1-OR,' ',XOR,'  ',1-XOR)
tabla_compuertas()
````


### Correspondencia completa con la clase del 26 de agosto
- BUFFER: conserva la entrada. NOT: la invierte.
- AND: 1 sólo si todas son 1. OR: 1 si al menos una es 1.
- NAND y NOR: negaciones de AND y OR; ambas son universales.
- XOR: 1 cuando las entradas son diferentes. XNOR: 1 cuando son iguales.

Las tablas de verdad siguientes verifican cada definición y permiten evaluar expresiones de dos y tres variables.


````python
def evaluar_compuertas(A,B):
    return {'BUFFER_A':A,'NOT_A':1-A,'AND':A&B,'OR':A|B,'NAND':1-(A&B),'NOR':1-(A|B),'XOR':A^B,'XNOR':1-(A^B)}
for A,B in product((0,1),repeat=2): print(A,B,evaluar_compuertas(A,B))
````

### Teoremas booleanos y ejercicios de simplificación

El álgebra de Boole permite simplificar expresiones lógicas y diseñar circuitos digitales con menos compuertas. En las expresiones siguientes,

**+**

representa OR,

**·**

representa AND y

**¬**

representa NOT.

#### Leyes fundamentales

| Ley | Expresión | Interpretación |
| --- | --- | --- |
| Identidad | A + 0 = A<br>A · 1 = A | Agregar el elemento neutro no cambia la variable. |
| Dominación | A + 1 = 1<br>A · 0 = 0 | En OR, 1 domina; en AND, 0 domina. |
| Idempotencia | A + A = A<br>A · A = A | Repetir una misma variable no altera el resultado. |
| Complemento | A + ¬A = 1<br>A · ¬A = 0 | Una variable y su negación producen siempre 1 en OR y 0 en AND. |
| Involución | ¬(¬A) = A | Negar dos veces devuelve el valor original. |
| Complemento de constantes | ¬0 = 1<br>¬1 = 0 | La negación invierte los niveles lógicos. |
| Conmutativa | A + B = B + A<br>A · B = B · A | El orden de las variables no afecta el resultado. |
| Asociativa | (A + B) + C = A + (B + C)<br>(A · B) · C = A · (B · C) | La forma de agrupar operaciones iguales no cambia el resultado. |
| Distributiva | A · (B + C) = A·B + A·C<br>A + B·C = (A + B)·(A + C) | Permite expandir o factorizar expresiones booleanas. |
| Absorción | A + A·B = A<br>A·(A + B) = A | Una variable absorbe el término que la contiene. |
| De Morgan | ¬(A + B) = ¬A · ¬B<br>¬(A · B) = ¬A + ¬B | Al negar un grupo se invierte la operación: OR pasa a AND y AND pasa a OR. |

#### Ejemplos de simplificación

1. A + A·B = A *(absorción)*
2. A·(B + C) = A·B + A·C *(distributiva)*
3. ¬(A·B) = ¬A + ¬B *(De Morgan)*
4. A + ¬A·B = A + B *(distributiva y complemento)*

**Consejo:**

Para simplificar, busca primero términos repetidos, complementarios o que puedan absorberse; después aplica distributiva o las leyes de De Morgan si hay negaciones sobre paréntesis.


### Leyes fundamentales
- Identidad: `A+0=A`, `A·1=A`
- Dominación: `A+1=1`, `A·0=0`
- Idempotencia: `A+A=A`, `A·A=A`
- Complemento: `A+¬A=1`, `A·¬A=0`
- Involución: `¬¬A=A`
- Conmutativa y asociativa
- Distributiva: `A(B+C)=AB+AC` y `A+BC=(A+B)(A+C)`
- Absorción: `A+AB=A`, `A(A+B)=A`
- De Morgan: `¬(AB)=¬A+¬B`; `¬(A+B)=¬A·¬B`

### Ejemplos de simplificación
1. `A + A·B = A` (absorción).
2. `A·B + A·¬B = A(B+¬B)=A`.
3. `¬(A+B) = ¬A·¬B` (De Morgan).
4. Una XOR puede expresarse como `¬A·B + A·¬B`.


````python
# Verificación exhaustiva de identidades por tabla de verdad
def verificar(nombre,izq,der,n=2):
    ok=all(bool(izq(*x))==bool(der(*x)) for x in product((0,1),repeat=n))
    print(nombre,':',ok)
verificar('Absorción A+AB=A',lambda A,B:A or (A and B),lambda A,B:A)
verificar('De Morgan NOT(A AND B)',lambda A,B:not(A and B),lambda A,B:(not A) or (not B))
verificar('XOR',lambda A,B:A^B,lambda A,B:((not A) and B) or (A and (not B)))
````

### NAND y NOR como compuertas universales

Con NAND: `NOT A = A NAND A`; `A AND B = NOT(A NAND B)`; `A OR B = (NOT A) NAND (NOT B)`.

Con NOR: `NOT A = A NOR A`; `A OR B = NOT(A NOR B)`; `A AND B = (NOT A) NOR (NOT B)`.

Esto demuestra que cualquier función construida con AND, OR y NOT puede implementarse usando exclusivamente NAND o exclusivamente NOR.


````python
def nand(a,b): return 1-(a&b)
def nor(a,b): return 1-(a|b)
def and_con_nand(a,b): return nand(nand(a,b),nand(a,b))
def or_con_nand(a,b): return nand(nand(a,a),nand(b,b))
def or_con_nor(a,b): return nor(nor(a,b),nor(a,b))
def and_con_nor(a,b): return nor(nor(a,a),nor(b,b))
for a,b in product((0,1),repeat=2):
    assert and_con_nand(a,b)==(a&b) and or_con_nand(a,b)==(a|b)
    assert and_con_nor(a,b)==(a&b) and or_con_nor(a,b)==(a|b)
print('Universalidad NAND y NOR verificada para todas las entradas')
````

### Leyes booleanas de las notas

Identidad, dominación, idempotencia, complemento, doble negación, complemento de constantes, conmutativa, asociativa, distributiva, absorción y De Morgan.

Método de simplificación: elimina constantes; agrupa términos comunes; aplica complemento; usa absorción; mueve negaciones con De Morgan; confirma el resultado por tabla de verdad.

Ejemplos: `A+AB=A`; `AB+A¬B=A`; `¬(AB)=¬A+¬B`; `¬(A+B)=¬A¬B`; XOR=`¬AB+A¬B`.


````python
identidades={
 'A+0=A':(lambda A,B:A or 0,lambda A,B:A),
 'A·1=A':(lambda A,B:A and 1,lambda A,B:A),
 'A+AB=A':(lambda A,B:A or (A and B),lambda A,B:A),
 'AB+A¬B=A':(lambda A,B:(A and B) or (A and not B),lambda A,B:A),
 'De Morgan AND':(lambda A,B:not(A and B),lambda A,B:(not A) or (not B)),
 'De Morgan OR':(lambda A,B:not(A or B),lambda A,B:(not A) and (not B))}
for nombre,(izq,der) in identidades.items():
    assert all(bool(izq(A,B))==bool(der(A,B)) for A,B in product((0,1),repeat=2))
    print(nombre,'✓')
````
